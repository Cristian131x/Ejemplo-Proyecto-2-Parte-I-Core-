# **Estaciones de Bicicletas - Dataset**

**Descripción:**

El dataset contiene registros de uso de bicicletas públicas junto con datos meteorológicos y temporales, facilitando el análisis de las variaciones en la demanda según las condiciones ambientales y los distintos momentos del día.

**Fuente:** Kaggle: https://www.kaggle.com/datasets/lakshmi25npathi/bike-sharing-dataset

Cada fila representa un registro de uso del sistema de bicicletas compartidas. El conjunto de datos combina información temporal, condiciones climáticas y cantidad de préstamos realizados.

**Variables principales:**
- Fecha y hora del registro
- Estación del año, mes y día de la semana
- Indicadores de feriados y días laborales
- Condiciones meteorológicas
- Temperatura y sensación térmica
- Humedad y velocidad del viento
- Cantidad de usuarios ocasionales
- Cantidad de usuarios registrados
- Total de bicicletas arrendadas

Estos datos permiten analizar los patrones de uso del sistema y cómo factores temporales y climáticos influyen en la demanda de bicicletas.

El conjunto de datos contiene **731 filas (registros diarios)** y **16 columnas (variables)**.
