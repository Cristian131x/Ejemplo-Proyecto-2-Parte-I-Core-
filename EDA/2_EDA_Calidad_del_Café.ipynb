# **Calidad del Café - Dataset**

**Descripción:**

El dataset reúne resultados de catas profesionales de café Arábica, integrando características organolépticas del grano con información sobre su origen geográfico, lo que permite analizar la relación entre la calidad sensorial y las condiciones de producción.

**Fuente**: Kaggle: https://www.kaggle.com/datasets/erwinhmtang/coffee-quality-institute-reviews-may2023

Cada fila representa una evaluación de un lote de café Arábica certificado. El conjunto de datos combina información sobre el origen del café, su proceso de producción y los resultados de catas profesionales.

**Variables principales:**
- País, región y finca de origen
- Productor y empresa responsable
- Altitud y año de cosecha
- Variedad y método de procesamiento
- Número y peso de los sacos
- Puntajes sensoriales (aroma, sabor, acidez, cuerpo, balance, dulzor, entre otros)
- Puntaje total de cata
- Humedad, color y calidad del grano verde
- Defectos identificados en los granos
- Información de certificación y evaluación

Estos datos permiten analizar la relación entre el origen, las características físicas y la calidad sensorial del café.

El conjunto de datos contiene **1.509 filas (evaluaciones de lotes de café)** y **42 columnas (variables)**.
